# Acessando a base de dados e criando arquivos históricos

- Links cvm

```python
url_itr = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/ITR/DADOS/"
url_dfp = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/DFP/DADOS/"
```

- Criando pastas do projeto

```python
pastas = [
    "data/raw/itr_0_downloaded",
    "data/raw/itr_1_extracted",
    "data/raw/dfp_0_downloaded",
    "data/raw/dfp_1_extracted",
    "data/interim/itr_2_concatenated",
    "data/interim/dfp_2_concatenated",
    "data/interim/itr_3_filtered",
    "data/interim/dfp_3_filtered"
]

for pasta in pastas:
    os.makedirs(pasta, exist_ok=True)
```

- Criando uma lista com o nome de todos os arquivos

```python
arquivos_itr_zip = []
arquivos_dfp_zip = []

for ano in range(2011, 2026):
    arquivos_itr_zip.append(f'itr_cia_aberta_{ano}.zip')
    
for ano in range(2010, 2026):
    arquivos_dfp_zip.append(f'dfp_cia_aberta_{ano}.zip')

print(arquivos_itr_zip)
print(arquivos_dfp_zip)
```

- Fazendo o download de todos os arquivos

```python
for arq in arquivos_itr_zip:
    destino = Path("data/raw/itr_0_downloaded") / arq
    # se já existe, pula
    if destino.exists():
        print(f"{arq} já existe")
        continue
    wget.download(
        url_itr + arq,
        str(destino)
    )
    print(f"Baixando {arq}")
        
for arq in arquivos_dfp_zip:
    destino = Path("data/raw/dfp_0_downloaded") / arq
    # se já existe, pula
    if destino.exists():
        print(f"{arq} já existe")
        continue
    wget.download(
        url_dfp + arq,
        str(destino)
    )
    print(f"Baixando {arq}")
```

- Extraindo os arquivos zip

```python
for arq in arquivos_itr_zip:
    ano = arq.replace("itr_cia_aberta_", "").replace(".zip","")
    arquivo_esperado = Path(f"data/raw/itr_1_extracted/itr_cia_aberta_DRE_con_{ano}.csv")
    # se já existe, pula
    if arquivo_esperado.exists():
        print(f"{arq} já extraído")
        continue
    with ZipFile(f"data/raw/itr_0_downloaded/{arq}","r") as zip_ref:
        zip_ref.extractall("data/raw/itr_1_extracted")
    print(f"Extraindo {arq}")
    
for arq in arquivos_dfp_zip:
    ano = arq.replace("dfp_cia_aberta_", "").replace(".zip","")
    arquivo_esperado = Path(f"data/raw/dfp_1_extracted/dfp_cia_aberta_DRE_con_{ano}.csv")
    # se já existe, pula
    if arquivo_esperado.exists():
        print(f"{arq} já extraído")
        continue
    with ZipFile(f"data/raw/dfp_0_downloaded/{arq}","r") as zip_ref:
        zip_ref.extractall("data/raw/dfp_1_extracted")
    print(f"Extraindo {arq}")
```

- Criando arquivos históricos (concatenação dos diferentes anos)

```python
nomes = ["", "BPA_con_", "BPA_ind_", "BPP_con_", "BPP_ind_", "composicao_capital_", "DFC_MD_con_", "DFC_MD_ind_", "DFC_MI_con_", 
             "DFC_MI_ind_", "DMPL_con_", "DMPL_ind_", "DRA_con_", "DRA_ind_", "DRE_con_", "DRE_ind_", "DVA_con_", "DVA_ind_", "parecer_"]

origem_itr = "data/raw/itr_1_extracted"
destino_itr = "data/interim/itr_2_concatenated"
os.makedirs(destino_itr, exist_ok=True)
for nome in nomes:
    arquivo_saida = (f"{destino_itr}/itr_cia_aberta_{nome}2011-2025.csv")
    # se já existe, pula
    if os.path.exists(arquivo_saida):
        print(f"Já existe: {arquivo_saida}")
        continue
    print(f"Processando ITR: {nome}")
    dfs = []
    anos = (
        range(2020, 2026)
        if nome == "composicao_capital_"
        else range(2011, 2026)
    )
    for ano in anos:
        arquivo = (f"{origem_itr}/itr_cia_aberta_{nome}{ano}.csv")
        if not os.path.exists(arquivo):
            print(f"Arquivo não encontrado: {arquivo}")
            continue
        df = pd.read_csv(
            arquivo,
            sep=";",
            decimal=",",
            encoding="ISO-8859-1"
        )
        dfs.append(df)
    if len(dfs) == 0:
        print(f"Nenhum arquivo encontrado para {nome}")
        continue
    arquivo_itr = pd.concat(dfs,ignore_index=True)
    arquivo_itr.to_csv(arquivo_saida,index=False)
    print(f"Salvo: {arquivo_saida} | {len(arquivo_itr):,} linhas")

# ------------------------------

origem_dfp = "data/raw/dfp_1_extracted"
destino_dfp = "data/interim/dfp_2_concatenated"
os.makedirs(destino_dfp, exist_ok=True)
for nome in nomes:
    arquivo_saida = (f"{destino_dfp}/dfp_cia_aberta_{nome}2010-2025.csv")
    # se já existe, pula
    if os.path.exists(arquivo_saida):
        print(f"Já existe: {arquivo_saida}")
        continue
    print(f"Processando DFP: {nome}")
    dfs = []
    anos = (
        range(2020, 2026)
        if nome == "composicao_capital_"
        else range(2010, 2026)
    )
    for ano in anos:
        arquivo = (f"{origem_dfp}/dfp_cia_aberta_{nome}{ano}.csv")
        if not os.path.exists(arquivo):
            print(f"Arquivo não encontrado: {arquivo}")
            continue
        df = pd.read_csv(
            arquivo,
            sep=";",
            decimal=",",
            encoding="ISO-8859-1"
        )
        dfs.append(df)
    if len(dfs) == 0:
        print(f"Nenhum arquivo encontrado para {nome}")
        continue
    arquivo_dfp = pd.concat(dfs,ignore_index=True)
    arquivo_dfp.to_csv(arquivo_saida,index=False)
    print(f"Salvo: {arquivo_saida} | {len(arquivo_dfp):,} linhas")
```

- Criando arquivos filtrados
    - Consolidado ou Individual?  
Quase sempre consolidado porque o consolidado representa o grupo econômico inteiro.
É isso que praticamente todos os analistas utilizam.
    - Csv ou Parquet?  
Parquet é melhor que o CSV para big data

```python
def convert_consolidated_to_parquet(origem, destino):
    os.makedirs(destino, exist_ok=True)
    origem = Path(origem)
    destino = Path(destino)
    for arquivo in origem.glob("*.csv"):
        # seleciona apenas arquivos consolidados
        if "_con_" not in arquivo.name:
            continue
        nome_parquet = (arquivo.name.replace("_con_", "_").replace(".csv", ".parquet"))
        arquivo_saida = destino / nome_parquet
        # se já existe, pula
        if arquivo_saida.exists():
            print(f"Já existe: {arquivo_saida.name}")
            continue
        print(f"Convertendo: {arquivo.name}")
        df = pd.read_csv(
            arquivo,
            sep=",",
            decimal=",",
            encoding="ISO-8859-1"
        )
        df.to_parquet(
            arquivo_saida,
            index=False,
            compression="snappy"
        )
        print(f"Salvo: {arquivo_saida.name} | {len(df):,} linhas")

# ITR
convert_consolidated_to_parquet(
    "data/interim/itr_2_concatenated",
    "data/interim/itr_3_filtered"
)

# DFP
convert_consolidated_to_parquet(
    "data/interim/dfp_2_concatenated",
    "data/interim/dfp_3_filtered"
)
```
