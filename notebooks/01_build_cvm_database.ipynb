# Acessando a base de dados e criando arquivos históricos

- Links cvm

```python
url_itr = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/ITR/DADOS/"
url_dfp = "https://dados.cvm.gov.br/dados/CIA_ABERTA/DOC/DFP/DADOS/"
```

- Criando pastas do projeto

```python
caminho_atual = Path.cwd()
pasta_anterior = caminho_atual.parent

pastas = [
    pasta_anterior / "data/raw/00_itr_downloaded",
    pasta_anterior / "data/raw/01_itr_extracted",
    pasta_anterior / "data/raw/00_dfp_downloaded",
    pasta_anterior / "data/raw/01_dfp_extracted",
    pasta_anterior / "data/interim/02_itr_concatenated",
    pasta_anterior / "data/interim/02_dfp_concatenated"
]

for pasta in pastas:
    os.makedirs(pasta, exist_ok=True)
```

- Criando uma lista com o nome de todos os arquivos

```python
arquivos_itr_zip = []
arquivos_dfp_zip = []

for ano in range(2011, 2026):
    arquivos_itr_zip.append(f'itr_cia_aberta_{ano}.zip')
    
for ano in range(2010, 2026):
    arquivos_dfp_zip.append(f'dfp_cia_aberta_{ano}.zip')

print(arquivos_itr_zip)
print(arquivos_dfp_zip)
```

- Fazendo o download de todos os arquivos

```python
for arq in arquivos_itr_zip:
    destino = Path(pasta_anterior / "data/raw/00_itr_downloaded") / arq
    # se já existe, pula
    if destino.exists():
        print(f"{arq} já existe")
        continue
    wget.download(
        url_itr + arq,
        str(destino)
    )
    print(f"Baixando {arq}")
        
for arq in arquivos_dfp_zip:
    destino = Path(pasta_anterior / "data/raw/00_dfp_downloaded") / arq
    # se já existe, pula
    if destino.exists():
        print(f"{arq} já existe")
        continue
    wget.download(
        url_dfp + arq,
        str(destino)
    )
    print(f"Baixando {arq}")
```

- Extraindo os arquivos zip

```python
for arq in arquivos_itr_zip:
    ano = arq.replace("itr_cia_aberta_", "").replace(".zip","")
    arquivo_esperado = Path(f"{pasta_anterior}/data/raw/01_itr_extracted/itr_cia_aberta_DRE_con_{ano}.csv")
    # se já existe, pula
    if arquivo_esperado.exists():
        print(f"{arq} já extraído")
        continue
    with ZipFile(f"{pasta_anterior}/data/raw/00_itr_downloaded/{arq}","r") as zip_ref:
        zip_ref.extractall(pasta_anterior / "data/raw/01_itr_extracted")
    print(f"Extraindo {arq}")
    
for arq in arquivos_dfp_zip:
    ano = arq.replace("dfp_cia_aberta_", "").replace(".zip","")
    arquivo_esperado = Path(f"{pasta_anterior}/data/raw/01_dfp_extracted/dfp_cia_aberta_DRE_con_{ano}.csv")
    # se já existe, pula
    if arquivo_esperado.exists():
        print(f"{arq} já extraído")
        continue
    with ZipFile(f"{pasta_anterior}/data/raw/00_dfp_downloaded/{arq}","r") as zip_ref:
        zip_ref.extractall(pasta_anterior / "data/raw/01_dfp_extracted")
    print(f"Extraindo {arq}")
```

- Criando arquivos históricos (concatenação dos diferentes anos)

```python
def adicionar_colunas_periodo(df: pd.DataFrame, nome_arquivo: str) -> pd.DataFrame:
    """
    Deriva granularidade trimestral a partir de DT_REFER (data a que o
    demonstrativo se refere) e preserva DT_RECEB (data de entrega/publicação
    à CVM), quando presentes no arquivo.

    - DT_REFER com mes 03/06/09 -> Q1/Q2/Q3 (ITR)
    - DT_REFER com mes 12       -> Q4 (equivalente ao DFP anual)
    - periodo: string "AAAAQn", chave única de junção ano+trimestre

    DT_RECEB é mantida como está (sem parsing aqui) para uso posterior no
    controle de look-ahead bias — a data de referência do balanço não é a
    mesma coisa que a data em que ele ficou publicamente disponível.
    """
    if "DT_REFER" not in df.columns:
        print(f"  [aviso] '{nome_arquivo}' não possui DT_REFER — colunas de período não adicionadas.")
        return df

    dt_refer = pd.to_datetime(df["DT_REFER"], errors="coerce")
    df["ano"] = dt_refer.dt.year
    df["mes_ref"] = dt_refer.dt.month
    df["trimestre"] = "Q" + (((df["mes_ref"] - 1) // 3) + 1).astype("Int64").astype(str)
    df["periodo"] = df["ano"].astype("Int64").astype(str) + df["trimestre"]

    if "DT_RECEB" not in df.columns:
        print(f"  [aviso] '{nome_arquivo}' não possui DT_RECEB — data de entrega/publicação indisponível.")

    return df


nomes = ["", "BPA_con_", "BPA_ind_", "BPP_con_", "BPP_ind_", "composicao_capital_", "DFC_MD_con_", "DFC_MD_ind_", "DFC_MI_con_", 
             "DFC_MI_ind_", "DMPL_con_", "DMPL_ind_", "DRA_con_", "DRA_ind_", "DRE_con_", "DRE_ind_", "DVA_con_", "DVA_ind_", "parecer_"]

origem_itr = pasta_anterior / "data/raw/01_itr_extracted"
destino_itr = pasta_anterior / "data/interim/02_itr_concatenated"
os.makedirs(destino_itr, exist_ok=True)
for nome in nomes:
    arquivo_saida = (f"{destino_itr}/itr_cia_aberta_{nome}2011-2025.parquet")
    # se já existe, pula
    if os.path.exists(arquivo_saida):
        print(f"Já existe: {arquivo_saida}")
        continue
    print(f"Processando ITR: {nome}")
    dfs = []
    anos = (
        range(2020, 2026)
        if nome == "composicao_capital_"
        else range(2011, 2026)
    )
    for ano in anos:
        arquivo = (f"{origem_itr}/itr_cia_aberta_{nome}{ano}.csv")
        if not os.path.exists(arquivo):
            print(f"Arquivo não encontrado: {arquivo}")
            continue
        df = pd.read_csv(
            arquivo,
            sep=";",
            decimal=",",
            encoding="ISO-8859-1"
        )
        dfs.append(df)
    if len(dfs) == 0:
        print(f"Nenhum arquivo encontrado para {nome}")
        continue
    arquivo_itr = pd.concat(dfs,ignore_index=True)
    arquivo_itr = adicionar_colunas_periodo(arquivo_itr, f"itr_cia_aberta_{nome}")
    arquivo_itr.to_parquet(arquivo_saida,index=False,compression="snappy")
    print(f"Salvo: {arquivo_saida} | {len(arquivo_itr):,} linhas")

# ------------------------------

origem_dfp = pasta_anterior / "data/raw/01_dfp_extracted"
destino_dfp = pasta_anterior / "data/interim/02_dfp_concatenated"
os.makedirs(destino_dfp, exist_ok=True)
for nome in nomes:
    arquivo_saida = (f"{destino_dfp}/dfp_cia_aberta_{nome}2010-2025.parquet")
    # se já existe, pula
    if os.path.exists(arquivo_saida):
        print(f"Já existe: {arquivo_saida}")
        continue
    print(f"Processando DFP: {nome}")
    dfs = []
    anos = (
        range(2020, 2026)
        if nome == "composicao_capital_"
        else range(2010, 2026)
    )
    for ano in anos:
        arquivo = (f"{origem_dfp}/dfp_cia_aberta_{nome}{ano}.csv")
        if not os.path.exists(arquivo):
            print(f"Arquivo não encontrado: {arquivo}")
            continue
        df = pd.read_csv(
            arquivo,
            sep=";",
            decimal=",",
            encoding="ISO-8859-1"
        )
        dfs.append(df)
    if len(dfs) == 0:
        print(f"Nenhum arquivo encontrado para {nome}")
        continue
    arquivo_dfp = pd.concat(dfs,ignore_index=True)
    arquivo_dfp = adicionar_colunas_periodo(arquivo_dfp, f"dfp_cia_aberta_{nome}")
    arquivo_dfp.to_parquet(arquivo_saida,index=False,compression="snappy")
    print(f"Salvo: {arquivo_saida} | {len(arquivo_dfp):,} linhas")
```


 - Consolidado ou Individual?

Ou seja, buscaremos prioritariamente nos arquivos "con", caso a empresa não seja identificada nestes arquivos, ela será procurada nos arquivos "ind".
